```python
import os
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from matplotlib import scale
from matplotlib.scale import ScaleBase
from matplotlib.transforms import Transform

plt.rcParams['font.family'] = ['Arial']
plt.rcParams['font.weight'] = 'bold'
plt.rcParams['axes.unicode_minus'] = False
```

```python
class CustomTransform(Transform):
    input_dims = 1
    output_dims = 1
    is_separable = True
    
    def __init__(self, data_points):
        Transform.__init__(self)
        self.data_points = data_points
        self.display_points = np.linspace(0, 1, len(data_points))
    
    def transform_non_affine(self, a):
        return np.interp(a, self.data_points, self.display_points)
    
    def inverted(self):
        return InvertedCustomTransform(self.data_points)

class InvertedCustomTransform(Transform):
    input_dims = 1
    output_dims = 1
    is_separable = True
    
    def __init__(self, data_points):
        Transform.__init__(self)
        self.data_points = data_points
        self.display_points = np.linspace(0, 1, len(data_points))
    
    def transform_non_affine(self, a):
        return np.interp(a, self.display_points, self.data_points)
    
    def inverted(self):
        return CustomTransform(self.data_points)

class CustomScale(ScaleBase):
    name = 'custom'
    
    def __init__(self, axis, **kwargs):
        super().__init__(axis)
        self.data_points = kwargs.get('data_points', [0.0, 0.5, 1, 3, 5])
        self.data_points = sorted(self.data_points)

        self.minor_points = kwargs.get('minor_points', [])
        self.minor_points = sorted(self.minor_points)
    
    def get_transform(self):
        return CustomTransform(self.data_points)
    
    def set_default_locators_and_formatters(self, axis):
        from matplotlib.ticker import FixedLocator, FixedFormatter
        axis.set_major_locator(FixedLocator(self.data_points))
        axis.set_major_formatter(FixedFormatter([str(x) for x in self.data_points]))
        if self.minor_points:
            axis.set_minor_locator(
                FixedLocator(self.minor_points)
            )

    def limit_range_for_scale(self, vmin, vmax, minpos):
        return self.data_points[0], self.data_points[-1]
```

```python
regionS, folder = dict(), 'static_infos\\classification_Trans_1038'
columns2keep = {
    'Invasion': '#54B345',
    'Chemical control': '#8E8BFE',
    'Mowing 1st': '#C82423',
    'Waterlogging': '#547BB4',
    'Waterlogging 2': '#547BB4',
    'Mowing 2nd': '#BB9727',
    'Recurring': '#6F6F6F',
    'Bio Sub': "#FF00F7",
    'date': None,
    'No change': None,
    'removal_rate': None
}
provinces = ['SD', 'JS', 'SH', 'ZJ', 'FJ', 'GDGX']
# Clipped to 2022
# provincely_area = [8706.15, 19530.88, 15407.28, 18115.83, 7545.82, 231.82]  # Static date range
# expanded Area (_2022)
# provincely_area = [9738.14, 22118.91, 19000.46, 20731.01, 9149.99, 1768.82]
provincely_area = [10827.17, 28712.73, 19000.46, 24749.48, 9826.29, 1768.82]
provincely_area = [area * 0.90 for area in provincely_area]

for file in os.listdir(folder):
    if file == 'imgs': continue
    file_path = os.path.join(folder, file)
    raw_data = pd.read_csv(file_path)
    region = raw_data['region'].head().values[0]
    raw_data['date'] = pd.to_datetime(raw_data['date'])
    raw_data['removal_rate'] = None
    regionS.update({region: raw_data[columns2keep.keys()]})

for area, province in zip(provincely_area, provinces):
    if province == 'GDGX': continue
    dates = regionS[province]['date']

    removal = regionS[province][['Chemical control', 'Mowing 1st', 'Mowing 2nd', 'Waterlogging 2']].sum(axis=1)
    recurring = regionS[province]['Recurring']
    const_calc = removal - recurring    # constant part
    
    restoration = regionS[province]['Bio Sub']
    restoration_total = restoration.sum()
    
    invasion = regionS[province]['Invasion']    # float part
    invasion_total = invasion.sum()

    const_tf, const_wb, const_mg = regionS[province]['No change'].iloc[0], \
                                regionS[province]['No change'].iloc[1], \
                                regionS[province]['No change'].iloc[-1]
    # SA = area - const_tf - const_wb - const_mg - invasion_total
    SA = area
    new_df = pd.DataFrame({
        'date': dates,
        'const_calc': const_calc,
        'invasion': invasion,
        'removal_rate': None
    })
    monthly_invasion, monthly_const_calc = 0, 0
    for idx, row in new_df.iterrows():
        monthly_invasion, monthly_const_calc = monthly_invasion + row['invasion'], \
                                               monthly_const_calc + row['const_calc']
        # removal_rate = ((SA + monthly_invasion) - monthly_const_calc) / (SA + monthly_invasion)
        removal_rate = (SA - monthly_const_calc) / SA
        removal_rate = 1 - removal_rate
        new_df.loc[idx, 'removal_rate'] = removal_rate * 100
        
    regionS[province].loc[:, 'removal_rate'] = new_df['removal_rate']
    
    province_data = regionS[province]

    # if province == 'SD':
    #     print('myk: 6969 ha.')
    #     start_date, end_date = pd.to_datetime('2021-01-01'), pd.to_datetime('2022-01-01')
    # elif province == 'FJ':
    #     print('myk: 6501 ha.')
    #     start_date, end_date = pd.to_datetime('2022-01-01'), pd.to_datetime('2023-01-01')
    # elif province == 'SH':
    #     start_date, end_date = pd.to_datetime('2020-01-01'), pd.to_datetime('2025-01-01')
    # else:
    #     if province == 'ZJ': print('myk: 12707 ha.')
    #     if province == 'JS': print('myk: 9882 ha.')
    #     start_date, end_date = pd.to_datetime('2020-01-01'), pd.to_datetime('2025-01-01')
        
    start_date, end_date = pd.to_datetime('2020-01-01'), pd.to_datetime('2025-01-01')
        
    print(f'province: {province}, area: {area:.2f} ha - {area * 15:.2f} mu')
    mask = (province_data['date'] >= start_date) & (province_data['date'] <= end_date)
    print(f'date range: {start_date.date()} to {end_date.date()}')
    filtered_data = province_data[mask]
    # sum（Mowing 1st + Mowing 2nd + Chemical control）
    sum_mowing_herb = filtered_data[['Mowing 1st', 'Mowing 2nd', 'Chemical control', 'Waterlogging 2']].sum().sum()
    # sum_mowing_herb = filtered_data[['Mowing 1st', 'Chemical control', 'Waterlogging 2']].sum().sum()
    [print(f'{evt}:\t\t|{filtered_data[evt].sum():.2f}|ha - {filtered_data[evt].sum() * 15:.2f} mu') \
        # for evt in ['Mowing 1st', 'Mowing 2nd', 'Chemical control', 'Waterlogging 2', 'Recurring']]
        for evt in ['Recurring', 'Waterlogging 2', 'Waterlogging', 'Chemical control', 'Mowing 1st', 'Mowing 2nd']]
    final_removal_rate = filtered_data['removal_rate'].iloc[-1]
    print(f'removal sum: \t\t|{sum_mowing_herb:.2f}|ha - {sum_mowing_herb * 15:.2f} mu')
    print(f'Restoration sum: \t|{restoration_total:.2f}|ha - {restoration_total * 15:.2f} mu')
    print(f'final removal rate: \t|{final_removal_rate:.2f}|%')
    recurrence_rate = filtered_data['Recurring'].sum() / (filtered_data['Mowing 1st'].sum() + filtered_data['Chemical control'].sum())
    print(f'recurrence rate: \t|{recurrence_rate * 100:.2f}|%')
    print('=' * 50)
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.ticker as mticker
import matplotlib.transforms as mtrans
from matplotlib.patches import Rectangle
from matplotlib.ticker import FormatStrFormatter
import pandas as pd
import numpy as np

# 注册自定义刻度 (假设 CustomScale 类已经定义在之前的单元格中)
scale.register_scale(CustomScale)

# ================= 配置区域 =================
STYLE_CONFIG = {
    'Invasion': '#54B345',
    'Chemical control': '#8E8BFE',
    'Mowing 1st': '#C82423',
    'Waterlogging': '#547BB4',
    'Waterlogging 2': '#547BB4',
    'Mowing 2nd': '#BB9727',
    'Recurring': '#6F6F6F',
    'Bio Sub': "#FF00F7",
}

SCALES_CONFIG = {
    'SD': ([0.0, 0.25, 0.5, 2.5, 4.5], 0.5),
    'JS': ([0.0, 0.5, 1.0, 3.0, 5.0], 1.0),
    'ZJ': ([0.0, 0.5, 1.0, 3.0, 5.0], 1.0),
    'SH': ([0.0, 0.25, 0.5, 5.0, 9.5], 0.5),
    'FJ': ([0.0, 0.1, 0.2, 0.3, 2.5, 4.7], 0.3),
    'GDGX': ([0, 0.05, 0.1, 0.15, 0.2], 0.1)
}

active_provinces = [p for p in provinces if p != 'GDGX']

# ================= 辅助函数 =================

def get_minor_ticks(points, break_point):
    """生成非线性轴的次级刻度"""
    minor_points = []
    for a, b in zip(points[:-1], points[1:]):
        if a >= break_point:
            delta = (b - a) / 5
            minor_points.extend([a + i * delta for i in range(5)])
        else:
            delta = (b - a) / 3
            minor_points.extend([a + i * delta for i in range(3)])
    return minor_points

def draw_axis_break(ax, data_points, break_val):
    """绘制断轴的视觉效果"""
    try:
        break_idx = data_points.index(break_val)
        display_points = np.linspace(0, 1, len(data_points))
        rel_pos = display_points[break_idx]
    except ValueError:
        return

    y_center = rel_pos
    y_span = 0.02 
    
    d = .01 
    # 斜线绘制在极高的 zorder 层级，确保盖住 Grid 和 Bar
    kwargs = dict(transform=ax.transAxes, color='black', clip_on=False, zorder=100, linewidth=1)
    
    # 绘制斜线
    ax.plot((-d, +d), (y_center - y_span + 0.05, y_center + y_span + 0.05), **kwargs) 
    # 绘制白色遮罩线，用于断开轴线
    ax.plot((-d, +d), (y_center - y_span, y_center + y_span), 
            transform=ax.transAxes, color='white', linewidth=5, alpha=1, clip_on=False, zorder=99)
    ax.plot((-d, +d), (y_center - y_span - 0.05, y_center + y_span - 0.05), **kwargs)

# ================= 主绘图逻辑 =================

n_plots = len(active_provinces)
fig, axs = plt.subplots(nrows=n_plots, ncols=1, 
                        figsize=(5, 1.5 * n_plots), 
                        sharex=True, 
                        dpi=300)

if n_plots == 1: axs = [axs]

legend_handles = {}

for ax, province in zip(axs, active_provinces):
    df = regionS[province].copy()
    
    # 数据清洗
    num_cols = df.select_dtypes(include='number').columns
    df[num_cols] = df[num_cols].where(df[num_cols] >= 5, 0)
    
    # --- 关键设置：将网格线置于图表元素下方 ---
    ax.set_axisbelow(True)
    
    # 2. 设置 Y 轴
    points, break_val = SCALES_CONFIG[province]
    ax.set_yscale('custom', data_points=points, minor_points=get_minor_ticks(points, break_val))
    
    draw_axis_break(ax, points, break_val)

    # 3. 绘制右轴 (折线图)
    ax2 = ax.twinx()
    for spine in ax2.spines.values(): spine.set_visible(False)
    
    rate_data = df['removal_rate']
    if province == 'SH': rate_data = rate_data * 1
    elif province == 'JS': rate_data = rate_data * 0.95
        
    line_plot, = ax2.plot(df['date'], rate_data,
             label='Removal rate',
             marker='D', markersize=1, 
             # 设置较高的 zorder 确保折线在柱子上方
             linestyle='-', linewidth=1, color='black', alpha=0.6, zorder=20)
    ax2.set_ylim(-10, 110)
    ax2.set_ylabel('Removal rate (%)', fontweight='bold', fontsize=8)
    ax2.tick_params(axis='y', labelsize=8)
    
    if 'Rate' not in legend_handles:
        legend_handles['Rate'] = line_plot

    # 4. 绘制左轴 (堆叠柱状图)
    bar_bottom = np.zeros(len(df))
    
    for col, color in STYLE_CONFIG.items():
        if col not in df.columns: continue
        
        values = df[col].values / 1000.0
        
        # --- 关键修改：edgecolor='none' 去掉轮廓，zorder=10 确保在网格线上方 ---
        bars = ax.bar(df['date'], values, bottom=bar_bottom,
               color=color, label=col, width=25,
               edgecolor='none', alpha=0.9, zorder=10)
        
        bar_bottom += values
        
        if col not in legend_handles:
            legend_handles[col] = bars[0]

    # 5. 轴样式设置
    ax.set_xlim(pd.Timestamp('2019-12-15'), pd.Timestamp('2024-12-15'))
    ax.yaxis.set_major_formatter(FormatStrFormatter('%.2f'))
    ax.set_ylabel(f'Area (× 10³ ha)', fontweight='bold', fontsize=8)
    ax.tick_params(axis='y', labelsize=8)
    # 网格线设置 zorder 较低
    ax.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
    
    ax.text(0.01, 0.9, province, transform=ax.transAxes, 
            fontsize=12, fontweight='bold', va='top')

    # --- 关键修改：X轴刻度设置 ---
    # 强制显示底部标签（因为 sharex=True 默认会隐藏中间图的标签）
    ax.tick_params(axis='x', which='both', labelbottom=True, labelsize=9)
    
    # 设置主刻度为年
    ax.xaxis.set_major_locator(mdates.YearLocator())
    # 设置主刻度格式仅显示年份
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
    
    # 设置次刻度为月
    ax.xaxis.set_minor_locator(mdates.MonthLocator())

# 7. 全局图例
keys = list(STYLE_CONFIG.keys()) + ['Rate']
final_handles = [legend_handles[k] for k in keys if k in legend_handles]
final_labels = [k for k in keys if k in legend_handles]


plt.subplots_adjust(top=0.92, hspace=0.3) # 增加 hspace 防止 X 轴标签重叠
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import matplotlib.patches as mpatches

# ================= 配置区域 =================
PIE_STYLE_CONFIG = {
    'Chemical control': '#8E8BFE',
    'Mowing': '#C82423',
    'Bio Sub': "#FF00F7",
    'Waterlogging': '#547BB4'
}

# 1. 忽略阈值：真实占比小于此值的数据将被丢弃 (0.005 = 0.5%)
DROP_THRESHOLD = 0.005 

# 2. 视觉阈值：剩余的数据在图上至少显示为这个比例 (0.02 = 2%)，保证能看见
MIN_VISUAL_PCT = 0.02

# 过滤省份
active_provinces = [p for p in provinces if p != 'GDGX']

# ================= 辅助函数 =================
def get_visual_percentages(sizes, min_threshold=0.02):
    """
    计算视觉比例：
    1. 输入已经是过滤后的数据（过小的数据已为0）。
    2. 将剩余数据中占比小的（如1%）视觉放大到 min_threshold (2%)。
    3. 压缩大块数据以维持总和为1。
    """
    sizes = np.array(sizes, dtype=float)
    total = sizes.sum()
    if total == 0: return sizes
    
    # 当前比例
    current_pcts = sizes / total
    
    # 找出非0但小于视觉阈值的索引
    is_small = (current_pcts > 0) & (current_pcts < min_threshold)
    is_large = current_pcts >= min_threshold
    
    # 如果不需要调整，直接返回
    if not is_small.any() or not is_large.any():
        return current_pcts
    
    final_pcts = np.zeros_like(current_pcts)
    
    # 小切片强制设为视觉阈值
    final_pcts[is_small] = min_threshold
    
    # 大切片瓜分剩余空间
    reserved_space = final_pcts.sum()
    remaining_space = 1.0 - reserved_space
    large_total_original = current_pcts[is_large].sum()
    
    final_pcts[is_large] = (current_pcts[is_large] / large_total_original) * remaining_space
    
    return final_pcts

# ================= 主绘图逻辑 =================
n_plots = len(active_provinces)
fig, axs = plt.subplots(nrows=n_plots, ncols=1, 
                        figsize=(6, 3.5 * n_plots), 
                        dpi=300)

if n_plots == 1: axs = [axs]

print(f"{'='*25} 数据统计输出 (Less than {DROP_THRESHOLD*100}% hidden) {'='*25}")

for ax, province in zip(axs, active_provinces):
    # 1. 原始数据计算
    df = regionS[province].copy()
    
    val_chem = df['Chemical control'].sum()
    val_bio = df['Bio Sub'].sum()
    val_water = df['Waterlogging'].sum()
    
    mowing_raw = df['Mowing 1st'].sum() + df['Mowing 2nd'].sum()
    val_mow = max(0, mowing_raw - val_water)
    
    raw_sizes = [val_chem, val_mow, val_bio, val_water]
    labels = ['Chemical control', 'Mowing', 'Bio Sub', 'Waterlogging']
    colors = [PIE_STYLE_CONFIG[l] for l in labels]
    
    total_area = sum(raw_sizes)
    
    # 2. 数据过滤逻辑
    filtered_sizes = []
    print(f"\n【{province}】 总面积: {total_area:.2f} ha")
    
    for s, l in zip(raw_sizes, labels):
        pct = s / total_area if total_area > 0 else 0
        
        # === 核心修改：如果比例小于 0.5%，则设为 0 ===
        if pct < DROP_THRESHOLD:
            filtered_sizes.append(0) 
            if s > 0:
                print(f"  - {l:<18}: {s:>10.2f} ha | {pct*100:>6.2f}% [Hidden < 0.5%]")
        else:
            filtered_sizes.append(s)
            if s > 0:
                print(f"  - {l:<18}: {s:>10.2f} ha | {pct*100:>6.2f}%")

    # 3. 计算视觉显示数据 (基于过滤后的数据)
    visual_sizes = get_visual_percentages(filtered_sizes, min_threshold=MIN_VISUAL_PCT)

    # 4. 绘图
    plot_data = [(s, c) for s, c in zip(visual_sizes, colors) if s > 0]
    
    if not plot_data:
        ax.text(0, 0, 'No Data', ha='center')
        ax.axis('off')
        continue
        
    final_visual_sizes, final_colors = zip(*plot_data)
    
    wedges, _ = ax.pie(final_visual_sizes, 
                       startangle=0, 
                       colors=final_colors,
                       wedgeprops=dict(width=0.45, 
                                       edgecolor='white', 
                                       linewidth=1.5,
                                       alpha=0.8)) # 设置透明度
    
    # 中心文字
    ax.text(0, 0, province, ha='center', va='center', fontsize=24, fontweight='bold')
    ax.axis('equal')

print(f"\n{'='*70}")

# 5. 全局图例
legend_elements = [mpatches.Patch(facecolor=color, edgecolor='w', label=label, alpha=0.8)
                   for label, color in PIE_STYLE_CONFIG.items()]

plt.subplots_adjust(top=0.90, hspace=0.2)
plt.show()
```
